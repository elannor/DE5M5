```python
df = pd.read_csv("Data/03_Library Systembook.csv")
df = df.dropna(how='all')
df['Customer ID'] = df['Customer ID'].astype('Int64')
df['Id'] = df['Id'].astype('Int64')
df['Book checkout'] = df['Book checkout'].astype(str).str.replace('32/05/2023','31/05/2023')
df['Book checkout'] = df['Book checkout'].astype(str).str.replace('"','')
df['Book checkout'] = pd.to_datetime(df['Book checkout'].str.replace('"','').str.strip(), format='%d/%m/%Y', errors='coerce')
df['Book Returned'] = pd.to_datetime(df['Book Returned'].str.replace('"','').str.strip(), format='%d/%m/%Y', errors='coerce')

def calculate_dates_difference(df, start_date, end_date, new_column):
    start_date = pd.to_datetime(df[start_date], errors='coerce')
    end_date = pd.to_datetime(df[end_date], errors='coerce')

    df[new_column] = (end_date - start_date).dt.days

    return df

df = calculate_dates_difference(
    df,
    start_date = "Book checkout",
    end_date = "Book Returned",
    new_column = "days_difference",
)

df.to_csv("Data/03_Library Systembook_Cleaned.csv", index=False)

df
```

```python
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def get_sum(self):
        return self.a + self.b
    
    def get_diff(self):
        return self.a - self.b
    
    def get_prod(self):
        return self.a * self.b
    
    def get_div(self):
        return self.a / self.b
    
if __name__ == "__main__":
    myCal = Calculator(a=3,b=2)
    print(myCal.get_sum())
```
